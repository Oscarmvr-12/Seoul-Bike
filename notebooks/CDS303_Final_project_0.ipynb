#### Dataset Description
This dataset contains individual-level Ttareungi (Seoul's public bike-share) rental records for January 1 - June 30, 2026 (18.5 million records), including rental/return times and stations, bike ID, trip duration and distance, and rider demographics.


```python
import pandas as pd
```

### English Translation Setup

The source file's column names are in Korean. The mapping and helper function below translate them to English so the notebook is readable to English-only readers.


```python
## English column-name mapping
COLUMN_TRANSLATIONS = {
    "자전거번호": "bike_id",
    "대여일시": "rental_datetime",
    "대여 대여소번호": "rental_station_number",
    "대여 대여소명": "rental_station_name",
    "대여거치대": "rental_dock",
    "반납일시": "return_datetime",
    "반납대여소번호": "return_station_number",
    "반납대여소명": "return_station_name",
    "반납거치대": "return_dock",
    "이용시간(분)": "usage_duration_min",
    "이용거리(M)": "usage_distance_m",
    "생년": "birth_year",
    "성별": "gender",
    "이용자종류": "user_type",
    "대여대여소ID": "rental_station_id",
    "반납대여소ID": "return_station_id",
}

def to_english(df):
    """Rename columns to English."""
    return df.rename(columns=COLUMN_TRANSLATIONS)
```

```python
## Load the full trip-level dataset (18.5M rows, ~3.2GB) - may take a minute or two
df = to_english(pd.read_csv("서울특별시 공공자전거 대여이력 정보_전체_2601-2606.csv", encoding="cp949"))
```
